### 2. Identifikasi Kolom PII (*Personally Identifiable Information*) & Quasi-Identifier
Berdasarkan pemeriksaan terhadap seluruh **26 kolom** pada `indonesia_pariwisata_super_prioritas_enriched.csv`[cite: 1]:
1. **Direct PII (Identitas Pribadi Langsung — Nama, NIK, Nomor Telepon, Email, Alamat Rumah):**
   * **Status:** **Tidak Ditemukan (0 kolom)**[cite: 1].
   * **Bukti:** Kolom `tourist_trip_id` hanya berisi kode anonim sintetis (contoh: `TOUR_IDN_2026_01618`) yang tidak terhubung dengan identitas sipil individu mana pun.
2. **Quasi-Identifier (Kombinasi Atribut yang Berpotensi Mengidentifikasi Kelompok/Individu):**
   * **Status:** **Ditemukan (5 kolom)**, yaitu: `observation_date` (tanggal kunjungan), `destination_cluster` (wilayah tujuan), `tourist_origin_category` (kategori asal wisatawan), `travel_party_size` (jumlah rombongan), dan `accommodation_type` (jenis akomodasi)[cite: 1].
   * **Mitigasi Risiko:** Karena tidak ada *Direct PII* dan atribut asal maupun akomodasi sudah digeneralisasi ke dalam kategori makro (contoh: `"Wisatawan Mancanegara Non-ASEAN"`, `"Hotel Bintang 4-5 Resort"`), risiko pelacakan balik terhadap individu spesifik (*re-identification risk*) sangat rendah dan aman digunakan untuk analisis agregat.

```python
import time
import json
import requests
from requests.adapters import HTTPAdapter
from urllib3.util.retry import Retry
from dotenv import load_dotenv

# 1. Memuat email kampus dari file .env untuk identitas User-Agent yang etis
load_dotenv()
EMAIL_KAMPUS = os.getenv("USER_AGENT_EMAIL", "nim_anda@instiki.ac.id")
HEADERS = {
    "User-Agent": f"INSTIKI-DAM-UTS-StudentProject/1.0 ({EMAIL_KAMPUS})"
}

# 2. Konfigurasi Session dengan mekanisme Retry otomatis
session = requests.Session()
retry_strategy = Retry(
    total=3,
    backoff_factor=1.5,
    status_forcelist=[429, 500, 502, 503, 504],
    allowed_methods=["GET"]
)
adapter = HTTPAdapter(max_retries=retry_strategy)
session.mount("https://", adapter)
session.mount("http://", adapter)

# 3. Daftar 8 Klaster Destinasi Pariwisata Indonesia (sama persis dengan dataset Soal 1)
klaster_destinasi = [
    {"destination_cluster": "Bali Greater Hub",          "query": "hotel Kuta Bali",        "lat": -8.7185, "lon": 115.1686},
    {"destination_cluster": "Borobudur Jawa Tengah",     "query": "hotel Borobudur Magelang", "lat": -7.6079, "lon": 110.2038},
    {"destination_cluster": "Danau Toba Sumatera Utara", "query": "hotel Parapat Danau Toba", "lat": 2.6611,  "lon": 98.9347},
    {"destination_cluster": "Mandalika Lombok NTB",      "query": "hotel Kuta Mandalika Lombok","lat": -8.8955,"lon": 116.2951},
    {"destination_cluster": "Labuan Bajo Flores NTT",    "query": "hotel Labuan Bajo",        "lat": -8.4964, "lon": 119.8877},
    {"destination_cluster": "Bromo Tengger Semeru",      "query": "hotel Bromo Probolinggo",  "lat": -7.9425, "lon": 112.9530},
    {"destination_cluster": "Raja Ampat Papua Barat",    "query": "resort Raja Ampat Waisai", "lat": -0.4316, "lon": 130.8157},
    {"destination_cluster": "Likupang Sulawesi Utara",   "query": "resort Likupang Minahasa", "lat": 1.6746,  "lon": 125.0561}
]

api_url = "https://photon.komoot.io/api/"
raw_json_responses = []
normalized_accommodations = []

print("=== MEMULAI AKUISISI DATA AKOMODASI PARIWISATA INDONESIA VIA API (SOAL 3) ===")

for klaster in klaster_destinasi:
    params = {
        "q": klaster["query"],
        "lat": klaster["lat"],
        "lon": klaster["lon"],
        "limit": 15 # Ambil maksimal 15 akomodasi per klaster (8 x 15 = hingga 120 baris)
    }

    try:
        # Mengirim request GET dengan User-Agent dan timeout 15 detik
        response = session.get(api_url, params=params, headers=HEADERS, timeout=15)

        if response.status_code == 200:
            data_json = response.json()
            data_json["destination_cluster"] = klaster["destination_cluster"]
            raw_json_responses.append(data_json)

            features = data_json.get("features", [])
            count_valid = 0

            # Normalisasi struktur GeoJSON bersarang ('features' -> 'properties' & 'geometry') menjadi tabel datar
            for feat in features:
                props = feat.get("properties", {})
                geom = feat.get("geometry", {})
                coords = geom.get("coordinates", [None, None])

                nama_akomodasi = props.get("name")
                negara = props.get("country", "")

                # Filter memastikan data berada di Indonesia dan memiliki nama akomodasi
                if nama_akomodasi and ("Indonesia" in negara or props.get("countrycode") == "ID"):
                    normalized_accommodations.append({
                        "osm_id": str(props.get("osm_id", "")),
                        "destination_cluster": klaster["destination_cluster"],
                        "nama_akomodasi": nama_akomodasi,
                        "jenis_tempat_osm": props.get("osm_value", "hotel"),
                        "kota_kabupaten": props.get("city", props.get("county", props.get("state", "Tidak Tercantum"))),
                        "provinsi": props.get("state", "Tidak Tercantum"),
                        "longitude": coords[0],
                        "latitude": coords[1]
                    })
                    count_valid += 1

            print(f"[OK - 200] {klaster['destination_cluster']}: ditemukan {count_valid} akomodasi.")
        else:
            print(f"[WARNING - {response.status_code}] Gagal mengambil data untuk {klaster['destination_cluster']}")

    except requests.exceptions.RequestException as err:
        print(f"[ERROR] Koneksi bermasalah pada {klaster['destination_cluster']}: {err}")

    # Jeda antar-request selama 1.5 detik
    time.sleep(1.5)

# 4. Menyimpan respons JSON mentah ke folder data/raw tanpa modifikasi
raw_api_file = "data/raw/photon_osm_akomodasi_indonesia_raw.json"
with open(raw_api_file, "w", encoding="utf-8") as f:
    json.dump(raw_json_responses, f, ensure_ascii=False, indent=2)

# 5. Mengubah hasil normalisasi menjadi DataFrame Pandas
df_api = pd.DataFrame(normalized_accommodations).drop_duplicates(subset=["osm_id", "nama_akomodasi"]).reset_index(drop=True)

print("\n=== HASIL NORMALISASI JSON DATA AKOMODASI (SOAL 3) ===")
print(f"Total Baris Data Akomodasi : {len(df_api)} baris (Syarat minimal soal: 30 baris)")
print(f"Total Kolom Data Akomodasi : {df_api.shape[1]} kolom")
print("\nDistribusi Jumlah Akomodasi per Klaster Destinasi:")
print(df_api["destination_cluster"].value_counts())

# 6. Mencatat Provenance Sumber 2 (API)
catatan_provenance = [p for p in catatan_provenance if "Sumber 2" not in p["sumber"]]
catatan_provenance.append({
    "sumber": "Sumber 2 - Photon OpenStreetMap Search API (Data Akomodasi Pariwisata Indonesia)",
    "url": api_url,
    "tanggal_akses": datetime.now().strftime("%Y-%m-%d %H:%M:%S WITA"),
    "lisensi": "Open Database License (ODbL 1.0) - OpenStreetMap & Komoot Photon",
    "file_mentah": raw_api_file,
    "md5_hash": hitung_md5(raw_api_file)
})

# Menampilkan 10 baris pertama
df_api.head(10)
```

```python
from urllib.robotparser import RobotFileParser
from bs4 import BeautifulSoup
from io import StringIO

target_url = "https://en.wikipedia.org/wiki/Tourism_in_Indonesia"
robots_url = "https://en.wikipedia.org/robots.txt"

print("=== 1. PEMERIKSAAN KEPATUHAN ROBOTS.TXT (SOAL 4) ===")
# Mengambil file robots.txt menggunakan session yang menyertakan User-Agent kampus
resp_robots = session.get(robots_url, headers=HEADERS, timeout=15)
print(f"HTTP Status robots.txt : {resp_robots.status_code}")

# Mem-parsing isi robots.txt menggunakan RobotFileParser
rp = RobotFileParser()
rp.parse(resp_robots.text.splitlines())

# Mengecek izin akses untuk User-Agent kita pada URL target
is_allowed = rp.can_fetch(HEADERS["User-Agent"], target_url)
crawl_delay = rp.crawl_delay(HEADERS["User-Agent"])

print(f"URL Target             : {target_url}")
print(f"User-Agent             : {HEADERS['User-Agent']}")
print(f"Status Kepatuhan       : {'DIIZINKAN (True)' if is_allowed else 'DILARANG (False)'}")
print(f"Crawl-Delay            : {crawl_delay if crawl_delay else 'Tidak ditentukan secara khusus (tetap gunakan jeda 2 detik)'}")

# Menampilkan cuplikan aturan User-agent: * dari robots.txt sebagai bukti audit
print("\nCuplikan aturan User-agent: * di robots.txt:")
lines = resp_robots.text.splitlines()
for idx, line in enumerate(lines):
    if line.strip().lower() == "user-agent: *":
        print("\n".join(lines[idx:idx+8]))
        break

# Jeda sopan 2 detik setelah membaca robots.txt sebelum mengambil halaman utama
time.sleep(2.0)

print("\n=== 2. PENGAMBILAN HALAMAN & EKSTRAKSI TABEL HTML STATIS ===")
df_scrape = pd.DataFrame()

if is_allowed:
    # Mengirim request GET ke halaman statis dengan User-Agent dan timeout 15 detik
    resp_wiki = session.get(target_url, headers=HEADERS, timeout=15)
    print(f"HTTP Status Halaman    : {resp_wiki.status_code}")
    resp_wiki.raise_for_status()

    # Menyimpan HTML mentah ke folder data/raw tanpa modifikasi sesuai Petunjuk No. 1
    raw_html_file = "data/raw/wikipedia_tourism_in_indonesia_raw.html"
    with open(raw_html_file, "w", encoding="utf-8") as f:
        f.write(resp_wiki.text)
    print(f"HTML mentah disimpan di: {raw_html_file}")

    # Parsing HTML menggunakan BeautifulSoup untuk mencari tabel statistik (class="wikitable")
    soup = BeautifulSoup(resp_wiki.text, "lxml")
    semua_tabel = soup.find_all("table", {"class": "wikitable"})
    print(f"Jumlah tabel 'wikitable' ditemukan: {len(semua_tabel)} tabel")

    if len(semua_tabel) > 0:
        # Membaca seluruh wikitable menggunakan pd.read_html
        daftar_df_wiki = pd.read_html(StringIO(str(semua_tabel)))

        # Mengambil tabel statistik kunjungan wisatawan (tabel pertama)
        df_scrape = daftar_df_wiki[0].copy()

        # Merapikan nama kolom jika terdapat MultiIndex atau catatan kaki [1], [2]
        if isinstance(df_scrape.columns, pd.MultiIndex):
            df_scrape.columns = [col[0] for col in df_scrape.columns]
        df_scrape.columns = [str(c).split("[")[0].strip() for c in df_scrape.columns]

        # Membersihkan tanda catatan kaki [1], [2] pada isi sel teks
        for col in df_scrape.select_dtypes(include="object").columns:
            df_scrape[col] = df_scrape[col].astype(str).str.replace(r"\[.*?\]", "", regex=True).str.strip()

        print(f"Ukuran tabel hasil scraping: {df_scrape.shape[0]} baris x {df_scrape.shape[1]} kolom")

        # Mencatat Provenance Sumber 3 (Web Scraping)
        catatan_provenance = [p for p in catatan_provenance if "Sumber 3" not in p["sumber"]]
        catatan_provenance.append({
            "sumber": "Sumber 3 - Wikipedia (Tourism in Indonesia - Visitor Statistics)",
            "url": target_url,
            "tanggal_akses": datetime.now().strftime("%Y-%m-%d %H:%M:%S WITA"),
            "lisensi": "Creative Commons Attribution-ShareAlike 4.0 (CC BY-SA 4.0)",
            "file_mentah": raw_html_file,
            "md5_hash": hitung_md5(raw_html_file)
        })
    else:
        print("Peringatan: Tidak ditemukan tabel wikitable pada halaman.")
else:
    print("Proses scraping dibatalkan karena tidak diizinkan oleh robots.txt.")

# Menampilkan hasil tabel yang berhasil di-scrape
df_scrape.head(15)
```
